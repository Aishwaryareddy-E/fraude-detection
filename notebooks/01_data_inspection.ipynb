# 01. Dataset Inspection & Structural Validation

**Official Project Title**: Financial Fraud Detection Model (Data Analytics – Streamlit Dashboard)

**Objective**: Rigorously verify dataset integrity, column uniformity, data types, missing values, duplicate records, and class distributions across both primary CSV files (`fraudTrain.csv` and `fraudTest.csv`) from `archive (8).zip`.

```
import pandas as pd
import numpy as np

# Load raw training and testing partitions
train_path = '../data/raw/fraudTrain.csv'
test_path = '../data/raw/fraudTest.csv'

train = pd.read_csv(train_path)
test = pd.read_csv(test_path)

print('Training Data Shape:', train.shape)
print('Testing Data Shape: ', test.shape)
```

```
# 1. Verify Column Parity
train_cols = train.columns.tolist()
test_cols = test.columns.tolist()

print('Matching columns count:', len(train_cols) == len(test_cols))
print('Columns match exactly:', train_cols == test_cols)
print('\nColumns:', train_cols)
```

```
# 2. Missing Values Check
print('Missing cells in Train:', train.isnull().sum().sum())
print('Missing cells in Test: ', test.isnull().sum().sum())
```

```
# 3. Duplicate Rows Check
print('Exact duplicate rows in Train:', train.duplicated().sum())
print('Exact duplicate rows in Test: ', test.duplicated().sum())
```

```
# 4. Fraud Label Distribution
print('=== Train Fraud Distribution ===')
print(train['is_fraud'].value_counts())
train_rate = (train['is_fraud'].sum() / len(train)) * 100
print(f'Train Fraud Rate: {train_rate:.4f}%')

print('\n=== Test Fraud Distribution ===')
print(test['is_fraud'].value_counts())
test_rate = (test['is_fraud'].sum() / len(test)) * 100
print(f'Test Fraud Rate: {test_rate:.4f}%')
```

```
# 5. Temporal Span Verification
print('Train Period:', train['trans_date_trans_time'].min(), 'to', train['trans_date_trans_time'].max())
print('Test Period: ', test['trans_date_trans_time'].min(), 'to', test['trans_date_trans_time'].max())
```
